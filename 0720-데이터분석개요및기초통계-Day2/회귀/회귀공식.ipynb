통계학에서 가장 기본이 되는 회귀 공식은 **선형 회귀(Linear Regression)** 공식

---

## 1. 단순 선형 회귀 (Simple Linear Regression) 공식

독립 변수($X$)가 **하나**인 가장 기본적인 회귀 모델입니다.

###  모형(Model)
종속 변수 $Y$와 독립 변수 $X$의 관계를 표현하는 이론적인 모형입니다.
$$Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i$$

* $Y_i$: $i$번째 관측값의 **종속 변수** (실제 값)
* $X_i$: $i$번째 관측값의 **독립 변수**
* $\beta_0$: **절편** (Intercept)
* $\beta_1$: **기울기** (Slope), $X$가 1단위 변할 때 $Y$가 변하는 정도 (회귀 계수)
* $\varepsilon_i$: **오차항** (Error Term), 모델이 설명하지 못하는 잔여 부분 (평균은 0, 분산은 $\sigma^2$인 정규분포를 따른다고 가정)

---

###  추정된 회귀식 (Estimated Regression Equation)
주어진 데이터($n$개의 표본)를 사용하여 모형의 계수($\beta_0, \beta_1$)를 추정한 식입니다.
$$\hat{Y}_i = \hat{\beta}_0 + \hat{\beta}_1 X_i$$

* $\hat{Y}_i$: $i$번째 관측값에 대한 **예측 값**
* $\hat{\beta}_0, \hat{\beta}_1$: **최소 제곱법(Least Squares Method)**으로 추정된 절편과 기울기 (회귀 계수)

###  회귀 계수 추정 공식 (최소 제곱 추정량, Least Squares Estimators)

**최소 제곱법**은 실제 값($Y_i$)과 예측 값($\hat{Y}_i$)의 차이(잔차) 제곱합을 최소화하는 $\hat{\beta}_0$와 $\hat{\beta}_1$을 찾는 방법입니다.

#### **기울기 $\hat{\beta}_1$ (Beta-one hat)**

$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2} = \frac{S_{XY}}{S_{XX}}$$

* 분자: $X$와 $Y$의 **공분산**의 합($S_{XY}$)
* 분모: $X$의 **분산**의 합($S_{XX}$)

#### **절편 $\hat{\beta}_0$ (Beta-zero hat)**

$$\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$$

* $\bar{Y}$: $Y$의 **표본 평균**
* $\bar{X}$: $X$의 **표본 평균**

> 이 공식은 추정된 회귀선이 반드시 표본의 평균점 ($\bar{X}, \bar{Y}$)을 통과한다는 성질을 이용한 것입니다.

---
##  회귀 공식 도출 과정 (최소 제곱법)

통계학에서 가장 기본적인 **단순 선형 회귀(Simple Linear Regression)** 모델의 회귀 계수($\hat{\beta}_0, \hat{\beta}_1$)는 **최소 제곱법(Ordinary Least Squares, OLS)**을 사용하여 도출됩니다. 이 방법은 실제 값과 예측 값 사이의 **잔차 제곱합(Sum of Squared Errors, SSE)**을 최소화하는 계수를 찾는 것을 목표로 합니다.

---

### 1. 회귀 모형과 잔차 정의

#### 1) 회귀식 (추정된 모형)
$n$개의 데이터 포인트 $(X_i, Y_i)$에 대해, 우리가 찾고자 하는 예측 모형은 다음과 같습니다.
$$\hat{Y}_i = \hat{\beta}_0 + \hat{\beta}_1 X_i$$

#### 2) 잔차 (Error)
잔차($e_i$)는 실제 관측값($Y_i$)과 모형의 예측값($\hat{Y}_i$)의 차이입니다.
$$e_i = Y_i - \hat{Y}_i = Y_i - (\hat{\beta}_0 + \hat{\beta}_1 X_i)$$


---

### 2. 잔차 제곱합 (SSE) 설정

최소 제곱법의 목적 함수(**Cost Function**)는 모든 잔차의 제곱을 합한 값(SSE)을 최소화하는 것입니다.

$$\text{SSE} = S(\hat{\beta}_0, \hat{\beta}_1) = \sum_{i=1}^{n} e_i^2 = \sum_{i=1}^{n} (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i)^2$$

우리의 목표는 이 $\text{SSE}$ 값을 **최소로 만드는** $\hat{\beta}_0$와 $\hat{\beta}_1$을 찾는 것입니다.

---

### 3. 최소화 조건 (편미분)

함수를 최소화하기 위해서는, 각 계수($\hat{\beta}_0$와 $\hat{\beta}_1$)에 대해 **편미분**하고 그 결과를 **0**으로 두어 극값을 찾습니다.

$$\frac{\partial S}{\partial \hat{\beta}_0} = \sum_{i=1}^{n} 2 (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) (-1) = 0$$

$$\frac{\partial S}{\partial \hat{\beta}_1} = \sum_{i=1}^{n} 2 (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) (-X_i) = 0$$


#### 1) $\hat{\beta}_0$에 대한 편미분
$$\frac{\partial S}{\partial \hat{\beta}_0} = \sum_{i=1}^{n} 2 (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) (-1) = 0$$

양변을 $-2$로 나누면,
$$\sum_{i=1}^{n} (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) = 0$$

합의 성질을 이용하여 각 항을 분리합니다.
$$\sum Y_i - \sum \hat{\beta}_0 - \sum \hat{\beta}_1 X_i = 0$$
$$\sum Y_i - n \hat{\beta}_0 - \hat{\beta}_1 \sum X_i = 0$$

$\hat{\beta}_0$에 대해 정리합니다.
$$n \hat{\beta}_0 = \sum Y_i - \hat{\beta}_1 \sum X_i$$
$$\hat{\beta}_0 = \frac{\sum Y_i}{n} - \hat{\beta}_1 \frac{\sum X_i}{n}$$

** 절편 $\hat{\beta}_0$ 공식 도출:**
$$\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$$
(여기서 $\bar{Y}$와 $\bar{X}$는 각각 $Y$와 $X$의 **표본 평균**입니다.)

---

#### 2) $\hat{\beta}_1$에 대한 편미분
$$\frac{\partial S}{\partial \hat{\beta}_1} = \sum_{i=1}^{n} 2 (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) (-X_i) = 0$$

양변을 $-2$로 나누면,
$$\sum X_i (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) = 0$$
$$\sum X_i Y_i - \hat{\beta}_0 \sum X_i - \hat{\beta}_1 \sum X_i^2 = 0$$

위에서 도출한 $\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$를 대입합니다.
$$\sum X_i Y_i - (\bar{Y} - \hat{\beta}_1 \bar{X}) \sum X_i - \hat{\beta}_1 \sum X_i^2 = 0$$
$$\sum X_i Y_i - \bar{Y} \sum X_i + \hat{\beta}_1 \bar{X} \sum X_i - \hat{\beta}_1 \sum X_i^2 = 0$$

$\hat{\beta}_1$을 포함하는 항과 포함하지 않는 항으로 나눕니다.
$$\sum X_i Y_i - n \bar{X}\bar{Y} = \hat{\beta}_1 \left( \sum X_i^2 - n \bar{X}^2 \right)$$
> 여기서 $\bar{Y} \sum X_i = \bar{Y} (n\bar{X}) = n \bar{X}\bar{Y}$ 입니다.

이것을 분산/공분산 형태로 변형하면 다음과 같습니다. (편차 사용)
$$\sum (X_i - \bar{X})(Y_i - \bar{Y}) = \hat{\beta}_1 \sum (X_i - \bar{X})^2$$

** 기울기 $\hat{\beta}_1$ 공식 도출:**
$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2}$$

---

### 4. 최종 회귀 계수 공식 요약

| 계수 | 공식 | 의미 |
| :--- | :--- | :--- |
| **기울기 $\hat{\beta}_1$** | $$\hat{\beta}_1 = \frac{S_{XY}}{S_{XX}}$$ | $X$와 $Y$의 **공분산**을 $X$의 **분산**으로 나눈 값 |
| **절편 $\hat{\beta}_0$** | $$\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$$ | 회귀선이 $(\bar{X}, \bar{Y})$를 통과하도록 설정하는 값 |

이 두 공식은 $\text{SSE}$를 최소화하는, 즉 데이터의 경향을 가장 잘 나타내는 **'최적의 선'**을 찾는 통계적 기반이 됩니다.

더 나아가 **다중 선형 회귀**의 행렬 공식 $(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{Y}$는 이 단순 회귀의 개념을 여러 개의 독립 변수로 확장한 것임을 이해할 수 있습니다.


## 2. 다중 선형 회귀 (Multiple Linear Regression) 공식

독립 변수($X$)가 **여러 개**($p$개)인 경우의 공식입니다.

###  모형 (Model)
$$Y_i = \beta_0 + \beta_1 X_{i1} + \beta_2 X_{i2} + \dots + \beta_p X_{ip} + \varepsilon_i$$

* $X_{i1}, X_{i2}, \dots, X_{ip}$: $p$개의 **독립 변수**
* $\beta_1, \beta_2, \dots, \beta_p$: 각 독립 변수에 대한 **회귀 계수** (기울기)

###  행렬 표기법 (Matrix Notation)

독립 변수가 많아지면 행렬을 사용하여 간단하게 표현합니다. 이는 계산이 복잡한 다중 회귀에서 필수적입니다.

$$\mathbf{Y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$$

* $\mathbf{Y}$: $n \times 1$ **종속 변수 벡터**
* $\mathbf{X}$: $n \times (p+1)$ **독립 변수 행렬** (첫 번째 열은 절편을 위한 1로 채워짐)
* $\boldsymbol{\beta}$: $(p+1) \times 1$ **회귀 계수 벡터** ($\beta_0, \beta_1, \dots, \beta_p$)
* $\boldsymbol{\varepsilon}$: $n \times 1$ **오차 벡터**

#### **회귀 계수 벡터 $\hat{\boldsymbol{\beta}}$ 추정 공식**

최소 제곱 추정량 $\hat{\boldsymbol{\beta}}$는 행렬 연산을 통해 다음과 같이 계산됩니다.

$$\hat{\boldsymbol{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{Y}$$

* $\mathbf{X}^T$: $\mathbf{X}$의 **전치 행렬**
* $(\mathbf{X}^T\mathbf{X})^{-1}$: $\mathbf{X}^T\mathbf{X}$의 **역행렬**

---

## 3. 회귀 모델 성능 평가 공식: 결정 계수 ($R^2$)

회귀 모델이 종속 변수 $Y$의 변동성을 얼마나 잘 설명하는지를 나타내는 지표입니다. 0과 1 사이의 값을 가지며, 1에 가까울수록 설명력이 높습니다.

$$R^2 = 1 - \frac{\text{SSE}}{\text{SST}} = \frac{\text{SSR}}{\text{SST}}$$

* **SST (Total Sum of Squares, 총 제곱합):** $Y$의 총 변동
    $$\text{SST} = \sum (Y_i - \bar{Y})^2$$
* **SSE (Error Sum of Squares, 오차 제곱합):** 모델이 **설명하지 못한** 변동 (잔차의 제곱합)
    $$\text{SSE} = \sum (Y_i - \hat{Y}_i)^2$$
* **SSR (Regression Sum of Squares, 회귀 제곱합):** 모델이 **설명한** 변동
    $$\text{SSR} = \sum (\hat{Y}_i - \bar{Y})^2$$


```python
import numpy as np

# 공부시간(X)
X = np.array([1, 2, 3, 4, 5])

# 시험점수(Y)
Y = np.array([50, 55, 65, 70, 80])

# -----------------------------
# 1. 평균 계산
# -----------------------------
X_bar = np.mean(X)
Y_bar = np.mean(Y)

# -----------------------------
# 2. Sxx, Sxy 계산
# -----------------------------
Sxx = np.sum((X - X_bar) ** 2)
Sxy = np.sum((X - X_bar) * (Y - Y_bar))

# -----------------------------
# 3. 회귀계수 계산
# β1 = Sxy / Sxx
# β0 = Y평균 - β1 * X평균
# -----------------------------
beta1 = Sxy / Sxx
beta0 = Y_bar - beta1 * X_bar

# -----------------------------
# 4. 예측값 계산
# -----------------------------
Y_hat = beta0 + beta1 * X

# -----------------------------
# 결과 출력
# -----------------------------
print(f"X 평균 = {X_bar:.2f}")
print(f"Y 평균 = {Y_bar:.2f}")

print(f"Sxx = {Sxx:.2f}")
print(f"Sxy = {Sxy:.2f}")

print(f"기울기 β1 = {beta1:.2f}")
print(f"절편   β0 = {beta0:.2f}")

print("\n예측값")
for x, y, pred in zip(X, Y, Y_hat):
    print(f"X={x}, 실제={y}, 예측={pred:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 공부시간
X = np.array([1, 2, 3, 4, 5])

# 시험점수(y)
y = np.array([50, 55, 65, 70, 80])

# 상관계수 계산
corr = np.corrcoef(X, y)
print(corr)


# 모델 생성
model = LinearRegression()
X = X.reshape(-1, 1) # 행렬변환
# 학습
model.fit(X, y)

print("기울기:", model.coef_[0])
print("절편:", model.intercept_)


# 공부시간이 6시간일 때
pred = model.predict([[6]])
print("공부시간이 6시간일 때 예측 점수:", pred)
```

```python
# 예측
y_pred = model.predict(X)

# -------------------------------
# 그래프
# -------------------------------
plt.figure(figsize=(6,4))

# 실제 데이터(산점도)
plt.scatter(X, y, s=80, label="Data")

# 회귀선
plt.plot(X, y_pred, color='r', linewidth=2, label="Regression Line")

plt.xlabel("Study Time")
plt.ylabel("Score")
plt.title("Linear Regression")

plt.legend()
plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_pred)
print("R² =", r2)
```

```python
from sklearn.metrics import (
    mean_squared_error,
    root_mean_squared_error,
    mean_absolute_error,
    r2_score
)

# 예측
y_pred = model.predict(X)

mse = mean_squared_error(y, y_pred)
rmse = root_mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"MSE  : {mse:.3f}")
print(f"RMSE : {rmse:.3f}")
print(f"MAE  : {mae:.3f}")
print(f"R²   : {r2:.3f}")
```
